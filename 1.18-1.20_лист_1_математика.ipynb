### 1.18-1.20.
Знакоопределенность и эквивалентность квадратичных форм исследуются на основе теорем Сильвестра. Положительная определенность требует строгой положительности всех главных угловых миноров, а отрицательная требует их чередования, начиная с отрицательного. По закону инерции две квадратичные формы эквивалентны над полем вещественных чисел тогда и только тогда, когда совпадают их сигнатуры (количества положительных и отрицательных собственных значений).


```python
# Задание 1.18
lam_val = sp.Symbol('lambda', real=True)
Q18 = sp.Matrix([
    [1, lam_val, 5],
    [lam_val, 4, 3],
    [5, 3, 1]
])
d1 = Q18[:1, :1].det()
d2 = Q18[:2, :2].det()
d3 = Q18.det()
print("Delta 1 =", d1)
print("Delta 2 =", d2)
print("Delta 3 =", sp.factor(d3))
print("Delta 3 всегда отрицательна, форма не может быть положительно определенной.")
```

```python
# Задание 1.19
Q19 = sp.Matrix([
    [lam_val, 1, -1],
    [1, -2, 1],
    [-1, 1, -3]
])
delta1 = Q19[:1, :1].det()
delta2 = Q19[:2, :2].det()
delta3 = Q19.det()
print("Delta 1 =", delta1)
print("Delta 2 =", delta2)
print("Delta 3 =", delta3)
sol19 = sp.solve([delta1 < 0, delta2 > 0, delta3 < 0], lam_val)
print(f"Ответ: {sol19}")
```

```python
# Задание 1.20
Q1_mat = sp.Matrix([
    [2, 4, -2],
    [4, 9, -5],
    [-2, -5, 3]
])
Q2_mat = sp.Matrix([
    [2, -2, -2],
    [-2, 3, 4],
    [-2, 4, 6]
])
signs_Q1 = [sp.sign(v) for v in Q1_mat.eigenvals().keys()]
signs_Q2 = [sp.sign(v) for v in Q2_mat.eigenvals().keys()]
print("Эквивалентны:", sorted(signs_Q1) == sorted(signs_Q2))
```
