### 1.1-1.3.
Для нахождения общего решения СЛАУ расширенная матрица системы приводится к приведенному ступенчатому виду методом Гаусса-Жордана. Столбцы с ведущими единицами соответствуют базисным переменным, которые затем линейно выражаются через свободные. Если система недоопределена (неизвестных больше, чем уравнений), можно аналитически исключить часть переменных путем алгебраических преобразований, чтобы найти прямую линейную зависимость между оставшимися переменными (например, выразить $y$ через $x$).


```python
# Задание 1.1
import numpy as np
import scipy.linalg as la
import sympy as sp
import matplotlib.pyplot as plt

sp.init_printing()
A=np.array([[1,5,-2,0],[-3,1,9,-5],[4,-8,-1,7]])
b=np.array([-7,9,0]).reshape(3,1)
A_b=np.hstack([A,b])
sp.Matrix(A_b).rref()
```

```python
# Задание 1.2
A2 = sp.Matrix([
    [5, -1, 5, 9, -2],
    [3,  0, 6, 9, -1],
    [2,  0, 4, 6, -1]
])
b2 = sp.Matrix([-3, 4, 0])

Ab2 = A2.row_join(b2)
print("а) rref(A|b) =")
display(Ab2.rref()[0])

vars2 = sp.symbols('x1:6')
sol_2 = sp.linsolve((A2, b2), vars2)

print("б) x =")
display(sol_2)
```

```python
# Задание 1.3
x, y, z, u = sp.symbols('x y z u')
eqs = [
    32*x + 97*y + 90*z + 48*u - 47,
    35*x + 84*y + 91*z + 54*u - 40,
    17*x + 18*y + 21*z + 13*u - 37
]
sol = sp.solve(eqs, (y, z, u))
y_expr = sp.simplify(sol[y])
a_const = y_expr.coeff(x, 0)
b_const = y_expr.coeff(x, 1)
print(f"y(x) = {y_expr}")
print(f"а = {a_const}, б = {b_const}")
```
