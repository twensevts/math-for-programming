### 1.13 и 1.14.
Собственные значения являются корнями характеристического уравнения матрицы. Минимальный многочлен аннулирует исходную матрицу и имеет наименьшую возможную степень среди всех таких многочленов, при этом его корнями выступают все собственные значения без учета алгебраической кратности. Он находится путем перебора комбинаций степеней множителей характеристического многочлена. Для каждого собственного значения можно найти и нормировать соответствующий собственный вектор.


```python
# Задание 1.13
A13 = sp.Matrix([[14, 10, 21], [12, 15, 26], [18, 11, 27]])
evals_evecs = A13.eigenvects()
all_pairs = []
for val, mult, vecs in evals_evecs:
    val_c = complex(val.evalf())
    vec_c = np.array(vecs[0].evalf(), dtype=complex).flatten()
    all_pairs.append((val_c, vec_c))
all_pairs.sort(key=lambda item: item[0].imag, reverse=True)
best_val, best_vec = all_pairs[0]
X_norm = best_vec / best_vec[0]
print(f"1) Im(Z) = {best_val.imag:.6f}")
print(f"2) Re(U) = {X_norm[1].real:.6f}")
```

```python
# Задание 1.14
import itertools
import sympy as sp

A14 = sp.Matrix([
    [-3, 7, -6, -1, 8, -3, -2],
    [1, -5, 8, -3, -5, 4, -5],
    [1, 3, -4, 6, -2, 0, 8],
    [-1, 11, -12, 9, 3, -3, 9],
    [-3, 11, -12, 5, 8, -4, 6],
    [-3, 2, 1, -4, 4, 1, -7],
    [1, -5, 6, -2, -4, 3, -2]
])

lam_sym = sp.Symbol('lambda')

char_poly = A14.charpoly(lam_sym).as_expr()
factors = sp.factor_list(char_poly)[1] 

powers = [range(1, m + 1) for f, m in factors]
combinations = list(itertools.product(*powers))
combinations.sort(key=sum)

min_poly = None

for p in combinations:
    test_poly = 1
    for i, (f, m) in enumerate(factors):
        test_poly *= f**p[i]
    
    expanded = sp.expand(test_poly)
    coeffs = sp.Poly(expanded, lam_sym).all_coeffs()
    
    degree = len(coeffs) - 1
    M_val = sp.zeros(7, 7)
    for i, c in enumerate(coeffs):
        M_val += c * (A14 ** (degree - i))
        
    if M_val == sp.zeros(7, 7):
        min_poly = test_poly
        break

min_poly
```
