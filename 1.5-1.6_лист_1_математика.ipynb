### 1.5 и 1.6.
Для матриц с заданными правилами формирования элементов (минимум, максимум или модуль разности индексов) определители вычисляются путем приведения к треугольному виду. Согласно теореме Адамара, определитель матрицы ограничен длинами ее векторов. Максимальный определитель достигается, когда матрица состоит только из элементов $1$ и $-1$ (нулевые элементы не увеличивают значение). Поиск оптимума для небольших матриц эффективно выполняется методом Монте-Карло.


```python
# Задание 1.5
n_test = 5
A_a = sp.Matrix(n_test, n_test, lambda i, j: min(i + 1, j + 1))
A_b = sp.Matrix(n_test, n_test, lambda i, j: max(i + 1, j + 1))
A_v = sp.Matrix(n_test, n_test, lambda i, j: abs((i + 1) - (j + 1)))
print("а) det =", A_a.det())
print("б) det =", A_b.det())
print("в) det =", A_v.det())
```

```python
# Задание 1.6
max_det = 0
best_M = None

for _ in range(100000):
    M = np.random.choice([-1, 1], size=(6, 6))
    d = abs(np.linalg.det(M))
    if d > max_det:
        max_det = round(d)
        best_M = M
        if max_det == 160:
            break

print(f"Максимальный определитель: {max_det}")
display(sp.Matrix(best_M))
```
