### 1.4.
Умножение матриц не обладает свойством коммутативности и выполняется по правилу умножения строки на столбец. Сложение матриц выполняется поэлементно и применимо только к матрицам одинаковой размерности.


```python
# Задание 1.4
M1 = np.array([[3, 0, 2], [0, 1, 3], [2, 2, 0], [0, 1, 0]])
M2 = np.array([[1, 2, -1, 2], [-2, 1, 1, 2], [2, 1, 1, 2]])
M3 = np.array([[0, -4, 6, 1], [2, 2, -5, -2], [2, -2, 6, 4], [1, 3, 0, 1]])
res = M1 @ M2 + M3
res
```
