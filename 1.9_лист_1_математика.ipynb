### 1.9.
Матричный ряд, представляющий собой геометрическую прогрессию, сходится при условии, что спектральный радиус матрицы меньше единицы. Предел бесконечной суммы такого ряда равен обратной матрице $(I - A)^{-1}$, умноженной на начальный элемент прогрессии.


```python
# Задание 1.9
gamma = 0.057
Amat = np.array([[7 * gamma, 14 * gamma], [9 * gamma, 1 * gamma]])
Imat = np.eye(2)
S_4_70 = sum(np.linalg.matrix_power(Amat, i) for i in range(4, 71))
print("1) max(S_4,70) =", S_4_70.max())
inv_IA = np.linalg.inv(Imat - Amat)
B_4 = inv_IA @ np.linalg.matrix_power(Amat, 4)
print("2) max(B_4) =", B_4.max())
for k_val in range(1, 200):
    Bk = inv_IA @ np.linalg.matrix_power(Amat, k_val)
    if Bk.max() < 0.0002:
        print("3) k =", k_val)
        break
```
