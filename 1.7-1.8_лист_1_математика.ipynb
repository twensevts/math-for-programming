### 1.7 и 1.8.
Прямое вычисление определителя или обратной матрицы для больших размерностей приводит к арифметическому переполнению или нехватке памяти. Определитель вычисляется как произведение диагональных элементов (через сумму логарифмов для обхода переполнения), а обратная матрица для треугольной остается треугольной, где элементы рассчитываются по точным формулам.


```python
# Задание 1.7
n7 = 80000
d = np.linspace(0.99, 1.01, n7)
log_diag_tail = np.sum(np.log(d[2:]))
det_top2 = d[0] * d[1] - 0.57 * 0.03
det_total = np.exp(np.log(det_top2) + log_diag_tail)
inv_top2_trace = (d[0] + d[1]) / det_top2
trace_inv = inv_top2_trace + np.sum(1.0 / d[2:])
print(f"1) det = {det_total:.3f}")
print(f"2) trace = {trace_inv:.3f}")
```

```python
# Задание 1.8
n8 = 43000
q = (1.07)**(1.0 / (n8 - 1))
d = 1.0 * (q ** np.arange(n8))
r = np.linspace(1.0, 1.10, n8)
b_diag = 1.0 / d
b_row0_rest = - r[1:] / (d[0] * d[1:])
min_elem = min(b_diag.min(), b_row0_rest.min())
sum_all = b_diag.sum() + b_row0_rest.sum()
print(f"1) min = {min_elem:.3f}")
print(f"2) sum = {sum_all:.1f}")
```
