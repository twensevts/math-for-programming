### 1.15-1.17.
По спектральной теореме любая симметричная вещественная матрица имеет ортогональный базис из собственных векторов, который приводится к ортонормированному виду процессом Грама-Шмидта. Это позволяет ортогональным преобразованием координат привести квадратичную форму к главным осям (сумме квадратов). Аналогичный алгоритм используется для классификации кривых второго порядка: матрица перехода задает поворот системы координат, а собственные значения становятся коэффициентами канонического уравнения (совпадение их знаков указывает на эллипс, а разные знаки на гиперболу).


```python
# Задание 1.15
A15 = sp.Matrix([[17, -8, 4], [-8, 17, -4], [4, -4, 11]])

ortho_vecs = []
evals = []
for val, mult, vecs in A15.eigenvects():
    ortho_vecs.extend(sp.GramSchmidt(vecs, orthonormal=True))
    evals.extend([val] * mult)
    
P_mat = sp.Matrix.hstack(*ortho_vecs)
D_mat = sp.diag(*evals)

display(D_mat)
display(P_mat)
```

```python
# Задание 1.16
A_b1 = sp.Matrix([[3, 4], [4, -3]])
A_b2 = sp.Matrix([[4, -2], [-2, 1]])

def get_ortho_PD(A):
    ortho_vecs = []
    evals = []
    for val, mult, vecs in A.eigenvects():
        ortho_vecs.extend(sp.GramSchmidt(vecs, orthonormal=True))
        evals.extend([val] * mult)
    return sp.Matrix.hstack(*ortho_vecs), sp.diag(*evals)

P1, D1 = get_ortho_PD(A_b1)
P2, D2 = get_ortho_PD(A_b2)

Q_ortho = sp.diag(P1, P2)
D_diag = sp.diag(D1, D2)

display(D_diag)
display(Q_ortho)
```
