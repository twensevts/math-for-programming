### 1.10.
Определитель матрицы с символьной переменной раскладывается по строкам или столбцам согласно теореме Лапласа. В результате получается алгебраический полином, из которого можно аналитически выделить числовые коэффициенты при соответствующих степенях неизвестной переменной.


```python
# Задание 1.10
x_sym = sp.Symbol('x')
Mmat = sp.Matrix([
    [7, 5, 5, x_sym],
    [2, 2, 6, 9],
    [3, x_sym, 8, 0],
    [3, 1, 1, 0]
])
poly_det = sp.expand(Mmat.det())
a_coeff = poly_det.coeff(x_sym, 2)
b_coeff = poly_det.coeff(x_sym, 1)
print("1) a =", a_coeff)
print("2) b =", b_coeff)
```
